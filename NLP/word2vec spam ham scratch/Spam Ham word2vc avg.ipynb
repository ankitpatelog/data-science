```python
import pandas as pd
message = pd.read_csv('SMSSpamCollection.txt' , sep='\t', names = ["label","message"])
```

```python
message
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
from nltk.corpus import stopwords

stopword_set = set(stopwords.words('english'))
corpus = []

for i in range(0, len(message)):
    # remove extra chars
    rem = re.sub('[^a-zA-Z]', ' ', message['message'][i])
    # make the sentence lower
    rem_lower = rem.lower()
    # split the sentence into words
    words = rem_lower.split()

    # apply stopwords and stemming together
    final = [lemmatizer.lemmatize(word) for word in words if word not in stopword_set]
    
    # join back into a single cleaned string and add to corpus
    corpus.append(' '.join(final))
```

```python
corpus
```

```python
for i, sent in enumerate(corpus):
    print(i,sent)
```

```python
from gensim.utils import simple_preprocess
# the simpl_preprocess takes the sentence and make it inot he list of words for onw snetence 
# where each index stores the words of one sentence
```

```python
words = []

for sent in corpus:
  words.append(simple_preprocess(sent))
```

```python
import numpy as np

def avg_word2vec(doc):
    # Collect vectors for words present in the model vocabulary
    # if word present in the model vocabulary then find its vector 
    vector_list =  [ model.wv[word] for word in doc if word in model.wv]

    # row wise mean
    return np.mean(vector_list,axis = 0)
```

```python
# now we are converting each word in the sentece mean
# after creatiing the word2vec model each word vecote is in the shape of the 100 rows and 1 col 
# so we can take the each word from sentence and transform into the one vector of dimension 100 in shape 100 rows and 1 col for one sent
# after that owr data is ready for model training

# also check the no of data points for x train and y train should be equal f
# if not equal make it equal by removng from the train data
```

```python
# now apply the fucntion to the sentence 
X = []
for i in tqdm(range(len(words))):
    X.append(avg_word2vec(words[i]))
```
