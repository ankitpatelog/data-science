```python
words = [
    "playing", "played", "plays", "player",
    "running", "runner", "runs", "ran",
    "studies", "studying", "studied", "study",
    "happiness", "happy", "happily",
    "connection", "connected", "connecting", "connects",
    "organization", "organize", "organized", "organizing",
    "relational", "relation", "relate", "related",
    "computing", "computer", "computation", "compute",
    "cats", "cat", "catlike",
    "fishing", "fished", "fisher", "fish",
    "driving", "driver", "driven", "drives",
    "walking", "walked", "walker", "walks",
    "beautiful", "beauty", "beautify",
    "creation", "creator", "creating", "created",
    "decision", "decide", "deciding", "decided",
    "movement", "moving", "moved", "moves",
    "national", "nation", "nationality",
    "friendship", "friends", "friendly",
    "careful", "carefully", "careless",
    "education", "educational", "educate", "educated",
    "development", "develop", "developing", "developed"
]
```

```python
# potter stemming 
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word + "--------->" + stemming.stem(word))
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import nltk

stemmer = PorterStemmer()
stop_words = set(stopwords.words('english'))

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])

    stemmed_words = [
        stemmer.stem(word)
        for word in words
        if word.lower() not in stop_words
    ]

    sentences[i] = ' '.join(stemmed_words)
```

```python
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
import nltk

stemmer = SnowballStemmer('english')
stop_words = set(stopwords.words('english'))

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])

    stemmed_words = [
        stemmer.stem(word)
        for word in words
        if word.lower() not in stop_words
    ]

    sentences[i] = ' '.join(stemmed_words)
```

```python
# we will find out the pas tags
# apply stopwords into all the words and then apply pos tag

import nltk
from nltk.corpus import stopwords

# Load stopwords only once
stop_words = set(stopwords.words('english'))

for sentence in new_sent:
    # Tokenize
    words = nltk.word_tokenize(sentence)

    # Remove stopwords
    final_words = [word for word in words if word.lower() not in stop_words]

    # Apply POS tagging
    pos_tags = nltk.pos_tag(final_words)

    print(pos_tags)
```
